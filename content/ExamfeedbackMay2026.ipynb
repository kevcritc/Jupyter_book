## 23.1 Section A  
Section A comprised of a set of questions which generally require the students to address a few points, draw a diagram with particular elements included, or perform a calculation through the manipulation of an equation.

### 23.1.1 Question 1
A straightforward electronic structure question.  Some people lost marks for incorrectly applying the auf bau principle and getting the order wrong. You should end with 5p$^1$ short hand was allowed. One marks was given for the correct number of electrons. 2 marks

### 23.1.2 Question 2 
Not everyone manage to get four points.  Only works for hydrogen atoms (1). Electrons treated as fixed orbitals (1). Fine structure not explained (1). No explination to chemical bonds (1).  Other options were accepted.

### 23.1.3 Question 3  
A scaling law question.  Part i was answered well. You needed to recognise this $f \propto \frac{1}{\sqrt{L}} $
Part ii) Was less well answered this time device lifetime is $\propto L$ because the wear is on the teeth of the cogs.

### 23.1.4 Question 4  
This was about semiconductors and their band gaps.  The important values were UV <320 nm and visible at 400 nm.  Therefore the band gap needed to fall in range of 3.1 to 3.9 eV. Marks were given for partial correct answers including the correct equation. An explination of absorption was needed to exaplin what happens (2 marks)

### 23.1.5 Question 5
This one troubles a lot of students.  A character takes up 36 pixels. A five letter word + a space is $6 \times 36 = 216$ pixels.

You could take any reasonable size for the pixel, but the idea what to think of the minimium (its a nanotech course), which is an atom - approximately $16 \text{Å}^2$ 
You can now calculate the area of a word.
Next the area of the pin head is $\pi r^2$
Divide the pin area by the word area to estimate the number of words. A value of about $8 \times 10^10$ or less was possible.
